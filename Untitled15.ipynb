```python
def temp_calculator(a,b,c,time):
  temp=a*time**2+b*time+c
  return temp
```

```python
#hard-coded variables
a=1.0
b=2.3
c=4.2
time=8.0
print("The temperature is",temp_calculator(a,b,c,time))
```

```python
#keyboard input
a=float(input("Enter a:"))
b=float(input("Enter b:"))
c=float(input("Enter c:"))
time=float(input("Enter time:"))
print("The temperature is",temp_calculator(a,b,c,time))
```

```python
#read from file
def read_from_file(filename):
  with open(filename) as f:
    lines=f.readlines()
  coeffiecients=[tuple(map(float,line.strip().split(','))) for line in lines]
  return coeffiecients
```

```python
#example: solve for a single set of inputs
coeffiecients=read_from_file('/content/input.txt')
a,b,c,time=coeffiecients[0]
temp=temp_calculator(a,b,c,time)
print("The temperature is",temp_calculator(a,b,c,time))
```

```python
#solve for all sets in the file
for idx,(a,b,c,time) in enumerate(coeffiecients):
  temp=temp_calculator(a,b,c,time)
  print("The temperature is",temp_calculator(a,b,c,time))
```
